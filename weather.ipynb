```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv("WeatherDataMerged.csv")
print(df.shape)
print(df.head(5))
orint(df.tail(5))
print(df.dtypes)
```

```python
duplicate_rows = df[df.duplicated()]
print(duplicate_rows.shape)

df.drop_duplicates(inplace=True)

print(df.isnull().sum())

df.dropna(subset=['temperature_2m', 'relativehumidity_2m', 'windspeed_10m'], inplace=True)
print(df.isnull().sum())
```

```python
df.describe()
```

```python
plt.figure(figsize=(14, 5))
sns.lineplot(x='time', y='temperature_2m', data=df)
plt.title("Temperature over time")
plt.xlabel("Date")
plt.ylabel("Temperature (°C)")
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(14, 5))
sns.lineplot(x='time', y='relativehumidity_2m', data=df)
plt.title("Relative humiditiy over time")
plt.xlabel("Date")
plt.ylabel("Humidity")
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(14, 5))
sns.lineplot(x='time', y='windspeed_10m', data=df)
plt.title("Wind speed over time")
plt.xlabel("Date")
plt.ylabel("Wind speed")
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(12, 5))
sns.boxplot(x=df['temperature_2m'])
plt.title("Temperature")
plt.show()
```

```python
plt.figure(figsize=(12, 5))
sns.boxplot(x=df['relativehumidity_2m'])
plt.title("relative humidity")
plt.show()
```

```python
plt.figure(figsize=(12, 5))
sns.boxplot(x=df['windspeed_10m'])
plt.title("wind speed")
plt.show()
```
